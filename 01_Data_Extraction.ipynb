```python
from googleapiclient.discovery import build
import pandas as pd

api_key = "AIzaSyCjDOL-rrbKj1lEI6oRn0tvnXJkQA-oNiE"
video_id = "nwJCZiLW3qo"

# Initialize the API client
youtube = build('youtube', 'v3', developerKey=api_key)

comments_data = []
next_page_token = None

# Extract up to 600 comments for the dataset
while len(comments_data) < 600:
    request = youtube.commentThreads().list(
        part='snippet',
        videoId=video_id,
        maxResults=100,
        pageToken=next_page_token
    )
    response = request.execute()

    for item in response.get('items', []):
        snippet = item['snippet']['topLevelComment']['snippet']
        comments_data.append({
            'Author': snippet['authorDisplayName'],
            'Comment': snippet['textOriginal'],
            'Likes': snippet['likeCount'],
            'Published At': snippet['publishedAt'],
            'Reply Count': item['snippet']['totalReplyCount']
        })

    next_page_token = response.get('nextPageToken')
    if not next_page_token:
        break

# Structure into a clean Pandas DataFrame
df = pd.DataFrame(comments_data)

# Display the first 5 rows to verify the data
display(df.head())

# Export the raw dataset
df.to_csv('structured_youtube_data.csv', index=False)
print("Data extraction complete and saved to CSV.")
```
