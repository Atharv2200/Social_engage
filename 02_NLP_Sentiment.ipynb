```python
import pandas as pd
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer

print("Step 1: Getting the sentiment analysis tools ready...")
nltk.download('vader_lexicon', quiet=True)

raw_csv_url = 'https://raw.githubusercontent.com/Atharv2200/Social_engage/refs/heads/main/structured_youtube_data_sample.csv'

print("Step 2: Loading your YouTube comments from GitHub...")
df = pd.read_csv(raw_csv_url)
print(f"Success! Loaded {len(df)} comments to analyze.\n")

sentiment_analyzer = SentimentIntensityAnalyzer()

relatable_triggers = [
    'relate', 'me too', 'feel this', 'exactly', 'so true',
    'struggle', 'anxiety', 'same', 'literally me'
]

def check_if_relatable(comment_text):
    if not isinstance(comment_text, str):
        return "Neutral"

    text_lower = comment_text.lower()

    if any(trigger in text_lower for trigger in relatable_triggers):
        return "Relatable"

    emotional_score = sentiment_analyzer.polarity_scores(comment_text)
    if emotional_score['compound'] <= -0.5 or emotional_score['compound'] >= 0.5:
         return "Relatable"

    return "Neutral"

print("Step 3: Reading through comments and tagging them as 'Relatable' or 'Neutral'...")
df['Sentiment_Tag'] = df['Comment'].apply(check_if_relatable)

print("\nHere is a sneak peek at how the model tagged your data:")
display(df[['Comment', 'Sentiment_Tag']].head(10))

print("\nStep 4: Saving the final results...")
df.to_csv('sentiment_tagged_data.csv', index=False)
print("Done! Your upgraded dataset 'sentiment_tagged_data.csv' is ready to download.")
```
