```python
import pandas as pd
import numpy as np
import scipy.stats as stats

raw_csv_url = 'https://raw.githubusercontent.com/Atharv2200/Social_engage/refs/heads/main/sentiment_tagged_data_sample.csv'
df = pd.read_csv(raw_csv_url)

df['Engagement_Score'] = df['Likes'] + (df['Reply Count'] * 2)

summary = df.groupby('Sentiment_Tag').agg(
    Total_Comments=('Comment', 'count'),
    Avg_Likes=('Likes', 'mean'),
    Avg_Replies=('Reply Count', 'mean'),
    Avg_Engagement=('Engagement_Score', 'mean')
).reset_index()

relatable_eng = df[df['Sentiment_Tag'] == 'Relatable']['Engagement_Score']
neutral_eng = df[df['Sentiment_Tag'] == 'Neutral']['Engagement_Score']

t_stat, p_value = stats.ttest_ind(relatable_eng, neutral_eng, equal_var=False)

print("=== VIRALITY & ENGAGEMENT METRICS ===")
display(summary)

print("\n=== HYPOTHESIS TESTING (T-TEST) ===")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print("Conclusion: Statistically significant! Relatable comments generate higher engagement.")
else:
    print("Conclusion: No statistically significant difference found in this sample.")

summary.to_csv('virality_metrics.csv', index=False)
```
